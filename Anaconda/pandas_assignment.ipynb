```python
import pandas as pd
import numpy as np
```

```python
#Q1 A
df = pd.read_csv("IRIS.csv")
print(df.head(10))
print(df.shape)
print(df.dtypes)
print(df.describe())
```

```python
#Q1 B
print(df[df["petal_length"] > 4.5])

print(df[df["species"] == "Iris-virginica"])
```

```python
#Q1 C
print(df.groupby("species").agg({
    "sepal_length": ["mean"],
    "petal_width": ["max"],
    "sepal_width": [np.std]
}))
```

```python
#Q1 D
df["petal_ratio"] = df["petal_length"] / df["petal_width"]
print(df.groupby("species")["petal_ratio"].mean())
```
