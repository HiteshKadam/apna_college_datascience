```python
#Real Dat

#Oscar Winner Movies
oscar_winners = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

plt.plot(years, oscar_revenue)
plt.title("Oscar Winner Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

plt.show()
```

```python
#Non-oscar Winner Movies
non_oscar_winners = ["Slumdog Millionaire", "Avatar", "Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 760, 829, 185, 275] # in $M



plt.plot(years, oscar_revenue,color='red',marker="v",linestyle="-.",linewidth=5,label="Oscar Winners")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Winners")

plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
plt.legend()

plt.show()
```

```python
plt.plot(years, oscar_revenue,label="Oscar Winners")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Winners")

plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
plt.legend()

# plt.style.use("ggplot")
plt.grid()

plt.savefig("oscar_vs_non_oscar.png")
plt.show()
```

```python
#Bar Charts

plt.bar(years, oscar_revenue,color="green")
plt.title("Oscar Winner Movie Revenue")
plt.xlabel("Year")
for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+5, str(oscar_revenue[i]), ha='center')
plt.ylabel("Revenue in $M")
plt.ylim(0, max(oscar_revenue)+100)

plt.show()

```

```python
import numpy as np
from wcwidth import width
#Compare with bar chart for non-oscar winners
x = np.arange(len(years))  # the label locations
width = 0.4

plt.bar(x-width/2, oscar_revenue, width=width, label="Oscar Winners") #Shifting to left
plt.bar(x+width/2, non_oscar_revenue, width=width, label="Non-Oscar Winners") #Shifting to right
plt.title("Oscar Winner Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
plt.legend()
```

```python
#Horizontal bar chart

plt.barh(oscar_winners,oscar_revenue)
plt.title("Oscar Winner Movie Revenue")
plt.xlabel("Revenue in $M")
plt.ylabel("Oscar Winner Movies")
```

```python
#Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E","Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]

plt.scatter(age, bp, s=bp, cmap="OrRd", c=bp)
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.title("Age vs Blood Pressure")
plt.colorbar(label="Blood Pressure")
plt.grid()

for i in range (len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+10)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]


plt.scatter(winter_temp, winter_humidity,label='Winter')
plt.scatter(summer_temp, summer_humidity, label='Summer')
plt.title("Summer vs Winter Data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")

plt.legend()
plt.show()
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

plt.pie(amounts, labels=expenses, autopct='%1.1f%%',
        wedgeprops={
            'edgecolor': 'black',
            "linewidth": 1.5,
            "linestyle": "--",
            "fill": True}, explode=(0, 0, 0, 0.2, 0), startangle=90, shadow=True)
plt.style.use("fast")
plt.show()
```
