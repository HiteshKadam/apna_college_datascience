```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style="darkgrid")
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
sns.relplot(x="total_bill",
            y="tip", 
            data=tips, 
            hue="smoker", 
            size="size" )
```

```python
sns.relplot(x="total_bill",
            y="tip", 
            data=tips,
            kind="line")
```

```python
sns.scatterplot(x="total_bill",
                y="tip",
                data=tips,
                hue="time")
```

```python
flights = sns.load_dataset("flights")

sns.lineplot(x="year",
             y="passengers",
                data=flights)
```

```python
#Bar plot
sns.barplot(x="day",
            y="tip",
            data=tips,
            hue="sex")
```

```python
#Box plot
sns.boxplot(x="day",
            y="total_bill",
            data=tips,
            hue="sex")
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(data=penguins, x="body_mass_g",bins=20)
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")

sns.heatmap(flights_pivot,
            cmap="coolwarm",
            annot=True,
            fmt="d",)

plt.title("Heatmap of Passengers by Month and Year")
```

```python
fig, ax = plt.subplots()

sns.lineplot(x="day", y="total_bill", data=tips, ax=ax, hue="sex",
             ci=False)

ax.set_title("Total Bill by Day")
ax.set_xlabel("Day of the Week")
ax.set_ylabel("Total Bill ($)")

fig.tight_layout()
```
