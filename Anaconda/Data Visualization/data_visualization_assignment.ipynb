```python
import matplotlib.pyplot as plt
```

```python
#Task

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25], #New York
    [18, 19, 17, 20, 21], #London
    [30, 32, 31, 33, 34], #Delhi
    [25, 26, 24, 27, 28] #Tokyo
]

fig, ax = plt.subplots(2,2, figsize=(12, 8))

city_no = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days, temperatures[city_no], marker='o')
        ax[i][j].set_title(cities[city_no])
        ax[i][j].grid(True)
        city_no += 1

fig.suptitle("Temperature Trends Over a Week in Different Cities", fontsize=16)
fig.supxlabel("Days of the Week", fontsize=12)
fig.supylabel("Temperature (°C)", fontsize=12)
fig.tight_layout()    
```
