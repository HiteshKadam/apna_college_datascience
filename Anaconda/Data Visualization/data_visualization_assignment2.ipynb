```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
#Q1
data = np.random.normal(loc=70, scale=10, size=200)
```

```python
plt.hist(data, bins=20, edgecolor='black')
plt.axvline(np.mean(data), color='red', linestyle='dashed', linewidth=1, label='Mean')
plt.show()
```

```python
#Q2
penguins = sns.load_dataset("penguins")
sns.relplot(x="flipper_length_mm", y="body_mass_g", hue="species", data=penguins)
```

```python
#Q3
tips = sns.load_dataset("tips")
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex")
```

```python
#Q4
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex",style="smoker")
plt.title("Scatter plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
#Q5
tips = sns.load_dataset("tips")
avg_tips = tips.groupby(['day', 'time'], observed=False)['tip'].mean().reset_index()
tips_pivot = avg_tips.pivot(index='day', columns='time', values='tip')
sns.heatmap(tips_pivot, annot=True, cmap="coolwarm")
```
