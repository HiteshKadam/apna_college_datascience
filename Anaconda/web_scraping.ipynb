```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
page_count = 1
while True:
    URL = URL = f"https://quotes.toscrape.com/page/{page_count}/"
    quotes_response = requests.get(URL)
    soup = BeautifulSoup(quotes_response.text,"lxml")
    quotes = soup.select("div.quote")
    if not quotes:
        print("No valid pages")
        break
    with open(f"scraped_data/quotes{page_count}.html","w") as f:
        f.write(quotes_response.text)
        print("Download from the page ", page_count)
    page_count+=1
```

```python
# Extract useful info
life_quotes = []
try:
    for i in range (1,page_count):
        with open(f"scraped_data/quotes{i}.html","r") as f:
            html_content = f.read()

        soup = BeautifulSoup(html_content,"lxml")

        all_quotes = soup.select("div.quote")


        for q in all_quotes:

            all_tags = []
            
            for t in q.select(".tags .tag"):
                all_tags.append(t.get_text())
            if "life" in all_tags:
                quote = q.select_one("span.text").get_text()
                author = q.select_one("small.author").get_text()
                life_quotes.append([quote,author])

    print(life_quotes)
except FileNotFoundError as e:
    print(e)
```

```python
df = pd.DataFrame(life_quotes,columns=["Quotes","Author"])
df.to_csv("cleaned_data/quotes.csv")
```
