```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
page_count = 1

while True:
    URL = f"https://www.scrapethissite.com/pages/forms/?page_num={page_count}"

    hockey_response = requests.get(URL)

    soup = BeautifulSoup(hockey_response.text,"lxml")

    data_count = soup.select_one("small").get_text().split()[0]
    if data_count == "0":
        print("This was Final Page!")
        break
    with open(f"scraped_data/hockey{page_count}.html","w") as f:
        f.write(hockey_response.text)
        print(f"Page {page_count} Done!")
        page_count+=1
```

```python
file_no = 1
teams = []
try:
    while True: 
        with open(f"scraped_data/hockey{file_no}.html","r") as f:
            html_content = f.read()

        soup = BeautifulSoup(html_content,"lxml")
        table_rows = soup.select_one("table.table").select("tr")
        for i in table_rows[1:]:
            row_data = []
            for td in i.select("td"):
                row_data.append(td.get_text(strip=True))
            teams.append(row_data)
        file_no += 1
except FileNotFoundError as e:
    print("All Files Complete")
```

```python
df = pd.DataFrame(teams, columns=["Team Name","Year","Win","Losses","OT Losses","Win %","Goals For(GF)","Goals Against(GA)","+/-"])
```

```python
df.to_csv("cleaned_data/hockey.csv")
```
